# Price Statistics

Berechnet Minimum, Mittelwert und Maximum der Spalte `price` aus `data/test_table.csv`.

Parameter `median` (`Yes`/`No`, Standard `No`): gibt zusätzlich den Median aus.

```python
try:
    dbutils.widgets.dropdown("median", "No", ["Yes", "No"], "Median Yes/No")
    show_median = dbutils.widgets.get("median") == "Yes"
except NameError:
    # Outside Databricks there is no dbutils
    show_median = False
```

```python
import pandas as pd

df = pd.read_csv("data/test_table.csv")
df
```

```python
price_min = df["price"].min()
price_mean = df["price"].mean()
price_max = df["price"].max()

print(f"Min:    {price_min:.2f}")
print(f"Mean:   {price_mean:.2f}")
print(f"Max:    {price_max:.2f}")

if show_median:
    print(f"Median: {df['price'].median():.2f}")
```
